```python
import numpy as np

data_file = r'C:\Users\Yazhini\Downloads\makeSenseOfCensus.csv'

data = np.genfromtxt(
    data_file,
    delimiter=',',
    skip_header=1
)

print(data)
```

```python
#How many rows and columns are present in the dataset?
print("Rows and columns:",data.shape)
```

```python
#Find the average age of all people
average_age = np.mean(data[:, 0])

print("Average age:", average_age)
```

```python
average_hours = np.mean(data[:, 6])

print("Average hours per week :", average_hours)
```

```python
average_income = np.mean(data[:, 7])

print("Average Income:", average_income)
```

```python
#Find the minimum and maximum age in the dataset.

min_age = np.min(data[:, 0])
max_age = np.max(data[:, 0])

print("Minimum age:", min_age)
print("Maximum age:", max_age)
```

```python
#Find the average number of years of education using education-num.

average_years = np.mean(data[:,1])
print("average number of years of education ",average_years)
```

```python
#Filtering and Conditions
#How many people are 40 years old or older?

people_older = np.sum(data[:,0] >= 40)

print("People aged 40 or older> 40:", people_older)
```

```python
#How many people have more than 10 years of education?

people_education = np.sum(data[:,1] > 10)

print("People have more than 10 years of education:", people_education)
```

```python
#How many people work more than 40 hours per week?

people_works = np.sum(data[:,6] > 40)

print("People work more than 40 hrs per week:", people_works)
```

```python
#How many people have a capital-gain greater than 0?

capital_gain = np.sum(data[:,4] > 0)

print("People have a capital-gain > 0", capital_gain)
```

```python
#How many people have a capital-loss greater than 0?

capital_loss = np.sum(data[:,5] > 0)

print("People have a capital-loss > 0", capital_loss)
```

```python
#Income Analysis
#How many people have an income greater than $50K?

highest_income = np.sum(data[:,7] > 50)

print("People have an income greater than $50K:", highest_income)
```

```python
#How many people have an income less than or equal to $50K?

lowest_income = np.sum(data[:,7] <= 50)

print("People have an income greater than $50K:", lowest_income)
```
